```python
# 랭체인으로 첫 챗봇 만들기
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model='gpt-4o-mini')
```

```python
# AIMessage, HumanMessage, SystemMessage 가 있음
from langchain_core.messages import HumanMessage
result = model.invoke([HumanMessage(content="안녕? 나는 홍길동이야")])
display(result)
```
