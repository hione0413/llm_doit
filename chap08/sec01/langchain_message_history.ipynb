```python
# 메세지 히스토리 연습
# 메모리 내에 메시지를 리스트형태로 보관
from langchain_core.chat_history import InMemoryChatMessageHistory
# 모델을 생성할 때 대화 기록을 함께 전달할 수 있도록 하는 클래스
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

model = ChatOpenAI(model="gpt-4o-mini")

# session id 를 기준으로 사용자의 대화 기록을 저장하는 딕셔너리
store = {}

# 세션 ID 에 따라 대화 기록을 가져오는 함수
def get_session_history(session_id: str):
    # 만약 해당 세션 ID가 store 에 없다면, 새로 생성해 추가
    if session_id not in store:
        # 메모리 내에 메시지를 리스트형태로 보관하는 객체 생성
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

# 모델 실행 시 대화 기록을 함께 전달하는 래퍼 객체 생성
with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

```python
config = {"configurable": {"session_id": "1234567890"}}

# response = with_message_history.invoke(
#     [HumanMessage(content="안녕? 나는 홍길동이야")],
#     config=config
# )

# print(response.content)

# invoke 대신 stream 사용하면 스트림 방식으로 쉽게 출력됨
for r in with_message_history.stream(
    [HumanMessage(content="안녕? 나는 홍길동이야")],
    config=config
):
    print(r.content, end="|")
```
