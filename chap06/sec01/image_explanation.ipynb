```python
# GPT 비전을 이용해 인터넷상의 이미지 설명 받기
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")

client = OpenAI(api_key=api_key)

message = [
    {
        "role": "user",
        "content": [
            {
                "type": "text",
                "text": "이 이미지에 대해 설명해 주세요."
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": "https://plus.unsplash.com/premium_photo-1753211477530-ac7d65a3119f?w=500&auto=format&fit=crop&q=60&ixlib=rb-4.1.0&ixid=M3wxMjA3fDB8MHxmZWF0dXJlZC1waG90b3MtZmVlZHw2fHx8ZW58MHx8fHx8"
                }
            }
        ]
    }
]

response = client.chat.completions.create(
    model="gpt-4o",
    messages=message
)

response
```

```python
# 로컬 이미지로 요청하기
# base 64 로 변환하여 요청
import base64

# 이미지를 인코딩하는 함수
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')

image_path = "C:/workspace/llm_study/01.test/chap06/sec01/image.jpg"

# 이미지를 base64 로 인코딩
base64_image = encode_image(image_path)

print(base64_image)
```

```python
# base 64 로 변환한 이미지 설명 요청하기
message = [
    {
        "role": "user",
        "content": [
            {
                "type": "text",
                "text": "이 이미지에 대해 설명해 주세요."
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": f"data:image/jpeg;base64,{base64_image}"
                }
            }
        ]
    }
]

response = client.chat.completions.create(
    model="gpt-4o",
    messages=message
)

response.choices[0].message.content
```
